```python
import math
Q = 5 # select 3 to 5 to begin with
f = 98e6 # Hz
R1 = 200
R2 = 50

ω = 2 * math.pi * f
X0 = R1/Q # not terminated into a load on X0 side
X1 = math.sqrt(R2/R1) * X0
X2 = X0 - X1

L0 = X0 / ω
C1 = 1 / (ω * X1)
C2 = 1 / (ω * X2)
print(f"L0 = {L0*1e9:.1f} nH")
print(f"C1 = {C1*1e12:.0f} pF")
print(f"C2 = {C2*1e12:.0f} pF")

0.1
```
