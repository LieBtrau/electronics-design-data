```python
# Peak-to-peak voltage to dBm conversion
import math

Vpp = 50e-3 # [V] Peak-to-peak voltage

Vp = Vpp / 2 # [V] Peak voltage
Vrms = Vp / math.sqrt(2) # [V] RMS voltage
Prms = Vrms**2 / 50 # [W] RMS power assuming 50 ohm load
dBm = 10 * math.log10(Prms / 1e-3) # [dBm] Power in dBm

print(f"Vpp: {Vpp*1e3} mV")
print(f"dBm: {dBm:.2f} dBm")
```
