# EDLC Capacitors

```python
# EDLC lifetime calculation
## Reference : 
# * https://english.jianghai.com/product_capacitor.html, select "Super Capacitors"
# * https://www.cde.com/resources/technical-papers/Supercapacitor-Technical-Guide.pdf

# Yageo FC0V474ZFTBR24
L1 = 1000 # Load life rating of the super capacitor (typically 1000 hours at rated temperature).
Tm = 85 # [°C] Maximum temperature rating of the supercapacitor.
Vr = 3.5 # [V] rated voltage of capacitor.

# Application
Ta= 65 # [°C] Ambient temperature the supercapacitor is going to be exposed to in the application.
Va = 3.4 # [V] applied voltage to capacitor

# L, expected life at operating condition.
L = L1 * 3.25**((Tm-Ta)/10) * 1.52**((Vr-Va)/0.1)
print(f"Capacitor life time at ambient : {L:.0f} h.")
```

```python
# Supercap lifetime calculation
## Reference : https://www.vishay.com/docs/28578/_3vedlcsdeliverlongeruseflife.pdf
ΔU0 = 0.2 # rule: voltage reduction of ΔU0 = 0.2 V doubles the lifetime
ΔT0 = 10 # rule: temperature reduction of ΔT0 = 10 K (°C) doubles the lifetime

# Component specs
t0 = 1000 # [h] specified lifetime
T0 = 85 # [°C] rated temperature
U0 = 3.5 # [V] rated voltage

# Application specs
T = 60 # [°C] applied temperature
U = 2.7 # [V] applied voltage

t = t0 * 2**((T0-T)/ΔT0) * 2**((U0-U)/ΔU0)
print(f"Capacitor life time at ambient : {t:.0f} h.")

# Convert life time hours to years
y = t / (24 * 365.25)
print(f"Capacitor life time at ambient : {y:.1f} years.")
```

```python
# Time keeping : needed capacitance
Δt = 3 * 86400 # [s], Requirement survival time : 3 days

Vcc = 3.3 # [V] supply voltage for supercap
Vf = 0.6 # [V] forward voltage of reverse current blocking diode
Vcap_max = Vcc - Vf
Vcap_min = 1.0 # [V]
I_RTC = 350e-9 # [A], Given the RV-4162 on Psi-SMARC
I_leak = 400e-9 # [A] (measured leakage current for SCAI0003)
I_tot = I_RTC + I_leak

ΔV = Vcap_max - Vcap_min
C = I_tot * Δt / ΔV
print(f"Capacitance : {C*1e3:.0f} mF.")
```

# Aluminium electrolytics

```python
# Component specs
L1 = 2000 # [h], load life rating
Vr = 35 # [V], rated voltage of the capacitor
Tm = 105 # [°C], maximum rated temperature of the capacitor
I_A = 0.5 # [A], actual ripple current at operating conditions

# Operating specs
Vo = 28 # [V], operating voltage
Ta = 50 # [°C], ambient temperature
ΔT = 5 # [°C], temperature rise due to ripple current
I_R = 0.5 # [A], rated ripple current at maximum rated temperature
```

```python
# Aluminium Electrolytical capacitor Jianghai

if Tm == 105:
    ΔT_0 = 5 # 105°C rated capacitor : 3.5 to 5
elif Tm == 85:
    ΔT_0 = 10 #85°C rated capacitor : 5 to 10, see databook.
else:
    print("Unrecognized Tm")
    exit(1)

if Tm == 105:
    if I_A > I_R:
        K_i = 4
    else:
        K_i = 2
elif Tm == 85:
    K_i = 2
else:
    print("Unrecognized Tm")
    exit(1)

L_0 = 2000 # [h], Lifetime under nominal load at upper category temperature (see catalogue)

# K_T : Temperature factor, according to Arrhenius equation, the lifetime of a capacitor doubles for every 10°C reduction in temperature.
K_T = 2**((Tm-Ta)/10)

# K_R : Ripple current factor
# Ripple current derating for frequency shouldn't be necessary for > 100 kHz.  ESR drops with frequency, so higher ripple currents are allowed
# at higher frequencies.
A = 1 - (I_A/I_R)**2
K_R = K_i**(A*ΔT_0/10)

# K_V : Voltage factor
K_V = 1 # For radial capacitors with a rated voltage < 160 V

# Total life time L
L = L_0 * K_T * K_R * K_V
print(f"Capacitor life : {L:.0f}h.")
```
